TYPES OF FUNCTION ARGUMENTS


1. POSITIONAL ARGUMENTS: Arguments assigned according to their positions

```python
def subtract(h,p):      #First argument is h and 2nd is p
    return(h-p)

print(subtract(1,3))
```

```python
def introduce(name, age):
    print(name, age)
introduce(name = "Prajakta", age = 29)
```

ARBITRARY POSITIONAL ARGUMENTS (args)
Use *args when you want to accept a variable number of positional arguments.

```python
def add_numbers(*numbers):    # "*" represents (*args) Tells python to accept all arguments user passes
  return sum(numbers)
print (add_numbers(1,3))
print (add_numbers(2,3,4,2))
```

ARBITRARY KEYWORDS ARGUMENTS(**kwargs) Used to accept variable number of "keyword" arguments

```python
def show_details(**details):
  print(details)
show_details (name = "Praj", age = 29, city = "Dundee")
```

Call a function from another Function

```python
def square(number):
  return number ** 2
def sum_of_squares (h,p):
  return square(h) + square(p)   #sum_of_square() function uses square function()
print(sum_of_squares(2,3))
```

Lamda Function - Is a anonymous function written as a single expression.

Regular Function

```python
def square(x):
  return x * x
```

Equivalent LAMbDA

```python
square = lambda x: x * x

print(square(5))
```

LAMDAS ARE OFTEN USEFUL FOR SHORT OPERATIONS SUCH AS SORTING

```python
student = [("Alex", 73), ("Praj", 98), ("Liam", 84)]
# Sort by the score (which is at index 1 of each tuple)
student.sort(key = lambda x: x[1])
print(student)
```

Recursive Function - A function that calls itself

```python
def factorial(n):
    if n <=1:
        return 1
    return n * factorial(n-1)
print(factorial(5))
```
